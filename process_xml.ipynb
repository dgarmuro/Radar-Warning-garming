```python
import json
```

```python
import xml.etree.ElementTree as ET
tree = ET.parse('gps_info.xml')
root = tree.getroot()
```

```python
ns = {"d": "http://datex2.eu/schema/1_0/1_0"}
```

```python
def get_name_by_type(loc, wanted_type):
    for name in loc.findall(".//d:name", ns):
        name_type = name.find("d:tpegDescriptorType", ns)
        value = name.find(".//d:descriptor/d:value", ns)

        if (
            name_type is not None
            and name_type.text == wanted_type
            and value is not None
        ):
            return value.text

    return None
```

```python
resultado = {}
```

```python
for loc in root.findall(".//d:predefinedLocation", ns):
    lat = loc.find(".//d:latitude", ns)
    lon = loc.find(".//d:longitude", ns)
    radar_id = loc.attrib.get("id")

    town = get_name_by_type(loc, "townName")

    if lat is not None and lon is not None:
        if not town in resultado: resultado[town] = [{"lat": float(lat.text), "long": float(lon.text)}]
        else: resultado[town].append({"lat": float(lat.text), "long": float(lon.text)})
        #print(radar_id, town, float(lat.text), float(lon.text))
```

```python
with open("radares_coord.json", "w", encoding = "utf-8") as f:
    json.dump(resultado, f, ensure_ascii=False, indent = 4)
```
